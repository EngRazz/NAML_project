# First RL problem of control:
The agent decide how moving the cart to not make the pole falling (due to gravity)
He receive a reward of 1 for every step where the pole is not falled (after taken an action movesx or moverx, the physics law implementation return the next position of the pole)
Here we show just a random choice of the action to take, so the pole doesn't stand for lot of time

```python
env = gym.make("CartPole-v1", render_mode="human") #creation of the environment, the render mode specifies how to visualize it (human --> window)

observation, info = env.reset() #env return the initial observation and additional information. It's the start of a new episode
#there are seed and options parameters to initializing it with specific values

print(f"Starting observation:{observation}")

episode_over = False
total_reward = 0

while not episode_over:
    action = env.action_space.sample() # action selection here is random from the action_space
    observation, reward, terminated, truncated, info = env.step(action) # execture the selected action --> environment updates and return new info / reward

    total_reward += reward
    episode_over = terminated or truncated
    time.sleep(0.5)  # rallenta
    
    # reward: +1 for each step the pole stays upright
    # terminated: True if pole falls too far (agent failed)
    # truncated: True if we hit the time limit (500 steps)

print(f"Episode finished! Total reward: {total_reward}")
env.close()
```

nb: An intelligent agent must map the observation (of an observation_space) into an action to be take (from the action_space) and not choosing randomly

They both are instances of the space python class, there are different kinds
and support different functions as contains() and sample()

```python
# Discrete action space (button presses)
env = gym.make("CartPole-v1")
print(f"Action space: {env.action_space}")  # Discrete(2) - left or right
print(f"Sample action: {env.action_space.sample()}")  # 0 or 1 (randomly sampled)

# Box observation space (continuous values)
print(f"Observation space: {env.observation_space}")  # Box with 4 values
# Box([-4.8, -inf, -0.418, -inf], [4.8, inf, 0.418, inf])
print(f"Sample observation: {env.observation_space.sample()}")  # Random valid observation
```

## Modifying the environment
with wrappers we can modify an env without alter the underlying code directly

Most environments created via gymnasium.make() will already be wrapped by default using TimeLimit (stops episodes after a maximum number of steps), OrderEnforcing (ensures proper reset/step order), and PassiveEnvChecker (validates your environment usage)

```python
from gymnasium.wrappers import FlattenObservation

# Start with a complex observation space
env = gym.make("CarRacing-v3")
print(env.observation_space.shape)
#(96, 96, 3)  # 96x96 RGB image

# Wrap it to flatten the observation into a 1D array
wrapped_env = FlattenObservation(env) #the observation space is reduced to a flat array thanks to the flattenobservation wrappers
print(wrapped_env.observation_space.shape)
#(27648,)  # All pixels in a single array

# This makes it easier to use with some algorithms that expect 1D input!!!
```

Common wrappers that beginners find useful:

TimeLimit: Issues a truncated signal if a maximum number of timesteps has been exceeded (preventing infinite episodes).

ClipAction: Clips any action passed to step to ensure it’s within the valid action space.

RescaleAction: Rescales actions to a different range (useful for algorithms that output actions in [-1, 1] but environment expects [0, 10]).

TimeAwareObservation: Adds information about the current timestep to the observation (sometimes helps with learning).
